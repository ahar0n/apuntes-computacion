# Ejericicios: Tratamiento y consulta de series temporales

## Contexto

Una estación meteorológica registra la temperatura (en grados Celsius) del aire cada hora, las 24 horas del día. Se requiere revisar sus datos, describir la evolución en el tiempo e identificar períodos con información incompleta.

Se dispone de observaciones desde el 1 de julio de 2026 a las 00:00 UTC hasta antes del 1 de octubre de 2026 a las 00:00 UTC.

La base de datos `db_temperatura` contiene:

- La tabla `temperatura`, con las columnas `instante_utc` y `temperatura_c`.
- La tabla `calendario`, con la secuencia completas horarios.
- La columna `instante_utc`, almacena fecha y hora en UTC y es clave primaria en ambas tablas. Se espera una observación por hora.
- La columna `temperatura_c` almacena valores decimales y admite `NULL`.

El archivo `temperatura.csv` contiene registros horarios de las temperaturas capturadas por la estación meteorológica.

## Ejercicio 2. Recuperar un período de la serie

La primera semana corresponde al intervalo `[1/jul 00:00, 8/jul 00:00)`, en UTC:

```sql
SELECT instante_utc,temperatura_c FROM temperatura
WHERE instante_utc>='2026-07-01 00:00:00'
  AND instante_utc<'2026-07-08 00:00:00'
ORDER BY instante_utc;
```

1. Ejecute la consulta y verifique el orden de las observaciones.
2. Explique por qué el 8 de julio a las 00:00 queda excluido. Cuente los registros y los valores disponibles de esta semana.
4. Adapte la consulta para la última semana: `[24/sep 00:00, 1/oct 00:00)`.

## Ejercicio 3. Construir una serie de temperaturas diarias

Ejecute la siguiente consulta:

```sql
SELECT DATE(instante_utc) AS dia,COUNT(*) AS registros,
       COUNT(temperatura_c) AS con_temperatura,
       ROUND(MIN(temperatura_c),3) AS minima_c,
       ROUND(MAX(temperatura_c),3) AS maxima_c,
       ROUND(AVG(temperatura_c),3) AS media_c
FROM temperatura
GROUP BY DATE(instante_utc) ORDER BY dia;
```

1. Interprete cada columna.
2. Determine la cantidad de días que aparecen y compárelos con el período completo.
3. Compare dos días con distinta cantidad de temperaturas disponibles.
4. Explique el significado del promedio de un día con datos incompletos.
5. Indique la información que se pierde al representar un día solo mediante su temperatura media.
