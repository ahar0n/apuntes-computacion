E2. Evalúe el cambio de estado de las variables durante la ejecución del siguiente programa.

```python
etiquetas = ["A12", "B07", "A05", "C21"]
cantidad_a = 0
iniciales = ""

for etiqueta in etiquetas:
    iniciales = iniciales + etiqueta[0]

    if etiqueta[0] == "A":
        cantidad_a = cantidad_a + 1

print(cantidad_a, iniciales)
```

E4. Antes de usar el intérprete de Python, ejecute:

1. ¿Qué valores alcanzará `i`?
2. ¿Qué resultado obtiene al realizar el seguimiento manual del estado de las variables?
3. Si existe un error, proponga una corrección.

```python
etiquetas = ["A12", "B07", "A05", "C21"]
i = 0
iniciales = ""

while i <= len(etiquetas):
    iniciales += etiquetas[i][0]
    i += 1

print(iniciales)
```
