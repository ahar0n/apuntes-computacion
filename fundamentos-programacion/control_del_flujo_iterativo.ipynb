# Control del flujo iterativo

Durante una iteración, el flujo normal consiste en repetir un conjunto de instrucciones mientras se mantenga una determinada condición o mientras existan elementos por procesar. Sin embargo, para resolver ciertos problemas resulta necesario modificar este comportamiento antes de que una iteración finalice de manera natural.

Python proporciona dos sentencias que permiten controlar de forma explícita el flujo iterativo normal: `break` y `continue`. La sentencia `break` permite finalizar anticipadamente la estructura iterativa, mientras que `continue` interrumpe únicamente la iteración actual y transfiere el control a la siguiente.

La utilización de estas sentencias __debe responder a una condición claramente identificable dentro del algoritmo__, de manera que contribuya a mantener un __flujo de ejecución comprensible y estructurado__.

## Finalización anticipada

La sentencia `break` permite realizar una finalización anticipada de una estructura iterativa. Cuando se ejecuta,

1. finaliza inmediatamente el bucle (`for` o `while`) más cercano que la contiene y
2. el flujo de ejecución continúa con la sentencia inmediatamente después del cuerpo del bucle.

En el siguiente ejemplo, la sentencia `break` permite que, una vez detectada la primera observación que supera el umbral, no es necesario continuar recorriendo la secuencia. El objetivo de este algoritmo es determinar si existe al menos una observación crítica y, en este caso, identificar la primera.

```python
observaciones = [18.4, 19.1, 20.3, 37.8, 21.0, 20.6]

umbral = 35.0
observacion_critica = None

for observacion in observaciones:
    if observacion > umbral:
        observacion_critica = observacion
        break

if observacion_critica is not None:
    print(f"Se detectó una observacion crítica: {observacion_critica} metros.")
else:
    print("No se detectaron observaciones críticas.")
```

## Continuación anticipada

La sentencia `continue` permite realizar una continuación anticipada de una iteración. Cuando se ejecuta,

1. se __omiten las sentencias restantes de la iteración actual__ y
2. la ejecución continúa con la siguiente iteración del bucle.

El iterador no termina, pero continúa con la siguiente iteración.

En el siguiente ejemplo, la sentencia `continue` se utiliza para el procesamiento de datos con registros inválidos.

En este caso, `-9999` representa un valor faltante o inválido. La sentencia `continue` permite omitir estos valores del registro y seguir procesando los demás. No es necesario terminar el iterador, porque todavía puede existir mediciones válidas más adelante.

```python
mediciones = [12.5, -9999, 13.1, 11.8, -9999, 14.0]

suma = 0
cantidad = 0

for medicion in mediciones:
    if medicion == -9999:
        continue

    suma += medicion
    cantidad += 1

if cantidad > 0:
    promedio = suma / cantidad
    print(f"Promedio de mediciones válidas: {promedio:.2f}")
else:
    print("No existen mediciones válidas.")
```

## Consideraciones de uso

- La documentación oficial de Python establece que `break` finaliza el bucle más interno que la contiene, mientras que `continue` omite el resto de la iteración actual y transfiere el control a la siguiente iteración [@python_simple_statements].
- Su utilización es apropiada cuando expresan de manera clara una condición de terminación anticipada o la necesidad de omitir parte del procesamiento de una iteración. Sin embargo, la presencia de múltiples sentencias `break` o `continue` dentro de un mismo bucle, especialmente cuando se combinan con estructuras condicionales anidadas, puede dificultar la comprensión del flujo de ejecución.
- El uso de `break` resulta especialmente natural cuando una búsqueda puede finalizar tan pronto como se encuentra el elemento deseado. De manera similar, `continue` puede ser útil cuando determinados elementos deben omitirse durante el recorrido de una colección sin finalizar por completo la estructura iterativa [@python_controlflow].
- Como principio general de estilo, las construcciones que modifican el flujo de ejecución deberían utilizarse de forma que mantengan explícita la intención del programa [@pep8].

> **Recomendación:** Utilizar `break` y `continue` cuando simplifiquen y hagan más evidente la lógica del algoritmo. Si su uso obliga a seguir múltiples rutas de ejecución para comprender el comportamiento del programa, considere reorganizar la solución.
