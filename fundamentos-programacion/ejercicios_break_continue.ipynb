CFI-001. Antes de usar el intérprete de Python, indique la salida que genera cada bloque de código.

````python
for numero in range(1, 8):
    if numero == 5:
        break
    print(numero)
````

````python
for numero in range(1, 8):
    if numero == 5:
        continue
    print(numero)
````

- ¿Qué diferencia existe entre las salidas?
- ¿Qué instrucción termina completamente el ciclo?
- ¿Qué instrucción permite que el ciclo continúe con otros valores?

CFI-009. El siguiente código fuente pretende mostrar los números del `1` al `10`, excepto el `5`, pero entra en un ciclo infinito:

````python
numero = 0
while numero < 10:
    numero += 1
    if numero == 5:
        numero -= 1
        continue
    print(numero)
````

- Explique por qué el ciclo no termina.
- Corrija el programa conservando el uso de la sentencia `continue`.
- Indique qué regla debe seguirse al usar `continue` dentro de un bucle `while`.
