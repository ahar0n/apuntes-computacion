En Python, las secuencias son __estructuras de datos que representan conjuntos finitos ordenados e indexados__, y que, en general, pueden accederse a sus elementos mediante índices. La [documentación oficial](https://docs.python.org/3/library/stdtypes.html#sequence-types-list-tuple-range) distingue tres tipos básicos de secuencias: `list`, `tuple` y `range`. Además, `str` es una secuencia especializada para texto.

| Tipo    | Mutabilidad | Contenido                  | Uso habitual                                     |
|:--------|:------------|:---------------------------|:-------------------------------------------------|
| `list`  | Sí          | Objetos de distintos tipos | Colecciones dinámicas                            |
| `tuple` | No          | Objetos de distintos tipos | Datos que no deben modificarse                   |
| `range` | No          | Números enteros            | Generación de secuencias numéricas e iteraciones |
| `str`   | No          | Caracteres Unicode         | Representación y procesamiento de texto          |

## Listas

Una lista es una secuencia [mutable](https://en.wikibooks.org/wiki/Python_Programming/Data_Types#Mutable_vs_Immutable_Objects) (o cambiable) ordenada de valores, que puede contener constantes o variables.

Las listas se pueden construir de varias maneras:

- Usando un par de corchetes para denotar la lista vacía: `[]`
- Usando corchetes, separando elementos con comas: `[a], [a, b, c]`
- Usando el constructor de tipo: `list()` o `list(iterable)`

Ejemplos de listas:

```python
numeros = [1, 2.1, 0.9764, -4, 5]
fruta = ['manzana', 'platano', 'naranja', 'pomelo']
```

## Tuplas

Las tuplas son secuencias [inmutable](https://en.wikibooks.org/wiki/Python_Programming/Data_Types#Mutable_vs_Immutable_Objects), que normalmente se utilizan para almacenar colecciones de datos heterogéneos, por ejemplo, fechas, direcciones, etc. También se usan para los casos en los que se necesita una secuencia inmutable de datos homogéneos.

Las tuplas se pueden construir de varias maneras:
- Usando paréntesis para denotar la tupla vacía: `()`
- Usando una coma final para una tupla [singleton](https://es.wikipedia.org/wiki/Singleton) (copia única): `a,` o `(a,)`
- Separación de elementos con comas: `a, b, c` o `(a, b, c)`
- Usando el constructor de tipo: `tuple()` o `tuple(iterable)`

Ejemplos de tuplas:

```python
aniversario = 'Diciembre', 25, 1983
tupla = ('a', 'b', 'c', 'd')
```

En general, el acceso a los elementos de una tupla se hace por medio del desempaquetado (_unpacking_) de una tupla. Para una tupla de `k` elementos, `k` variables pueden ser asignadas a cada uno de los elementos.

```python
mes, dia, anio = aniversario
print(f'El aniversario es en el año {anio}, {mes} {dia}.')
```

## Cadenas de texto

Las cadenas de texto (texto o datos textuales) se manejan en Python con objetos de tipo `str`, que es un tipo inmutable de secuencia de código [Unicode](https://en.wikipedia.org/wiki/Unicode). Los literales de cadena se pueden crear de varias maneras:

- Usando comillas simples: `'permite comillas "dobles" incrustadas'`
- Usando comillas dobles: `"permite comillas 'individuales' incrustadas"`
- Usando comillas triples permite abarcar varias líneas: `'''Tres comillas simples'''`, `"""Tres comillas dobles"""`.

Los literales de cadena que son parte de una sola expresión y solo tienen espacios en blanco entre ellos se convertirán implícitamente en un solo literal de cadena. Es decir, `("buenos" "días") == "buenos días"`.

Las cadenas también se pueden crear a partir de otros objetos (conversión) usando el constructor `str`.

## Operadores de secuencias

- Los operadores de secuencias permite realizar algunas operaciones entre secuencias. Algunos operadores básicos:

| Operación | Símbolo |
| :--       | :--:|
| Anidar    | + |
| Repetir   |  ∗ |
| Pertenencia  | in |

La operación anidar `+` __permite concatenar secuencias__:

```python
['manzana'] + ['repollo', 'zapallo']
```

```python
'radio' + '2'
```

```python
('a', 'b') + ('c',)
```

La __concatenación debe aplicarse sobre secuencias del mismo tipo__, en caso contrario, se genera un error de tipo `TypeError`:

```python
'manzana' + ['repollo', 'zapallo']
```

La operación repetir `*` __permite concatenar copias de la misma secuencias__:

```python
['manzana'] * 3
```

Notar que el tipo de dato derivado de una operación de secuencia, es del mismo tipo de la secuencia que se está operando:

```python
type('auto' + 'movil')
```

```python
type('la' * 4)
```

El operador `in` __permite comprobar si un valor o variable pertenece a una secuencia__.
- Es un operador binario con asociatividad por la derecha.

```python
'manzana' in fruta
```

- Puede ser utilizado en conjunto con el operador `not` lo que permite comprobar si elemento no pertenece a una secuencia.

```python
'manzana' not in fruta
```

## Longitud y posición de elementos

Cuando la longitud de una secuencia es `n`, cada elemento tiene como índice los números `0, 1, ..., n-1`. 

La __longitud de una secuencia__, se puede obtener por medio de la función `len()` que devuelve el número de elementos de una secuencia:

```python
len(fruta)
```

El ítem `i` de la secuencia es seleccionado de la forma `[i]`.

```python
'manzana'[2]    # tercer elemento
```

```python
fruta[-1]       # último elemento
```

```python
fruta[3]        # cuarto elemento
```

Para conocer la __posición de un valor__ que existe en una secuencia, la función `index()` devuelve el índice del valor que pertenece a una secuencia:

```python
fruta.index('platano')  # indice de un elementos en una lista
```

```python
'platano'.index('a')    # indice de un elementos en una cadena de carácteres
```

Si el valor está duplicado, la función entrega la primera ocurrencia.

```python
'platano'.index('a')
```

Además, la función `index()` permite indicar la posición de ínicio de búsqueda de la ocurrencia del valor sobre una secuencia. 

Por ejemplo, en la siguiente expresión, la función devuelve la ocurrencia del valor `'o'` encontrado en el segmento `cadena[4:]` de la secuencia `'automovil'`:

A continuación se presenta una sentencia que identifica la segunda ocurrencia del caracter <code>'o'</code> sobre la secuencia `cadena`:

```python
cadena = 'automovil'
cadena.index('o', cadena.index('o') + 1)
```

## Segmentación

Las secuencias también soportan la __segmentación__ `[i:j]`. Esta operación, selecciona todos los elementos con índice `k` tal que `i <= k < j`. 

Cuando se utiliza como una expresión, el segmento es una secuencia del mismo tipo, lo que implica que el conjunto de índices se vuelve a numerar para que comience en `0`.

Por ejemplo, aplicando segmentación sobre un `str`:

```python
palabra = 'aeropuerto'
palabra[:4]
```

En Python, los índices negativos sirven para contar elementos de una secuencia (como listas o textos) desde el final hacia el principio. El último elemento tiene el índice `-1`, el penúltimo el `-2`, y así sucesivamente. Por ejempplo:

```python
palabra[len(cadena)-5:]
```

Otro ejemplo, obteniendo segmentos de secuencias utilizando la ocurrencia de elementos:

```python
palabra[0: cadena.index('o') + 1]
```

Aplicando la segmentación sobre una `list`:

```python
letras = ['a', 'b', 'c', 'd']
letras[:2]
```

## Mutación de secuencias

- Python proporciona un conjunto operaciones que permiten modificar, agregar y eliminar valores de una secuencia mutable.
- Para modificar elementos de una secuencia, se selecciona dicho elemento mediante su índice y se le asigna el nuevo valor.

Para __agregar elementos la final de una secuencia__, se emplea la función `append()`. En el siguiente ejemplo, esta función se utiliza para agregar el elemento `amarillo` a la lista `colores`:

```python
colores = ['blanco', 'rojo', 'negro']
colores.append('amarillo')
```

Para __insertar un nuevo valor__ en la posición cuyo índice es `k` se utiliza la operación `insert()`. Por ejemplo, para insertar un elemento en la cuarta posición:

```python
colores.insert(2, 'verde')
```

Para __eliminar un valor__ de una lista se utiliza la función `remove()`. Si el valor a eliminar está repetido, se elimina sólo su primera aparición:

```python
colores.remove('rojo')
```

## Asignación de referencias

- En Python, __las listas son almacenadas mediante referencias a las secuencias de elementos__.
- La asignación del contenido de una variable que almacena una lista, supone la copia de la referencia hacia dichos elementos de la lista, almacenados en una zona de memoria. Una interpretación gráfica:

Notar que la comparación entre las distintas variables entrega como resultado que son iguales:

```python
(a == b) and (a == c) and (b == c)
```

Para estos casos, cuando dos (o más) variables hacen referencia a la misma zona en memoria, el operador `is` __permite identificar si dos objetos residen en la misma zona de memoria__.

```python
c is a
```

```python
a is c
```

```python
a is b
```

Notar que, al realizar una __asignación de una variable a partir de referencias, los elementos de dicha variable no son independientes__. Esto es, si se modifica la variable, entonces se modifican los elementos a los cuales hacen referencia.

```python
a.append(8)
print(c)
```

## Iterar sobre secuencias

Es posible iterar sobre una secuencia utilizando estructuras de control repetitivas.

Por ejemplo, usando un bucle `while`:

```python
palabra = 'computador'
i = 0
while i < len(palabra):
    print(palabra[i])
    i += 1
```

Se puede obtener un resultado equivalente utilizando un ciclo `for`:

```python
palabra = 'computador'
for i in range(len(palabra)):
    print(palabra[i])
```

En el siguiente ejemplo, en cada iteración del ciclo `for` la variable de control es asignada con un elemento de la secuencia en cada iteración.

```python
cadena = 'colegio'
for i in cadena:
    print(i)
```
