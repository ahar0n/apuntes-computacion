# Diccionarios

## Definición

Un **diccionario** es una colección de asociaciones entre claves y valores, son identificados como tipos de mapeo o _mapping types_ [@psf314diccionarios]. Cada **clave** identifica una entrada y permite acceder a su **valor** asociado. La entrada es el par formado por ambos. En Python, el tipo `dict` proporciona esta representación [@psf314estructuras].

Las listas y las tuplas son secuencias y su acceso elemental utiliza posiciones enteras. Un **diccionario utiliza claves para establecer las asociaciones**, sin embargo, una clave entera no representa necesariamente una posición [@psf314estructuras].

Las claves deben ser **hashables**. Un objeto hashable tiene un valor hash estable durante su existencia y una comparación de igualdad compatible con ese valor. Los objetos iguales deben tener el mismo hash. El hash es un entero que interviene en la organización de las consultas, no es un identificador necesariamente único. Cadenas y enteros son hashables. Las listas y los diccionarios no lo son. Una tupla solo es hashable si todos sus elementos también lo son [@psf314glosario].

Por tanto, **inmutable no constituye por sí solo un criterio suficiente para definir una clave**. Por ejemplo, la tupla `("P01", 2026)` puede utilizarse como clave, pero la tupla `("P01", [2026])` contiene una lista y no puede utilizarse.

Cada clave identifica un solo valor. Los **valores pueden repetirse y pueden pertenecer a tipos diferentes** [@psf314estructuras]. Que Python admita esa variedad no significa que cualquier combinación sea adecuada.

## Creación e inserción de elementos

Un diccionario puede construirse enumerando sus pares entre llaves. `{}` y `dict()` crean diccionarios vacíos [@psf314estructuras].

Ejemplos de creación de diccionarios:

````python
existencias = {"P01": 12, "P02": 0}
inventario_vacio = {}
otro_inventario_vacio = dict()
````

La asignación a una clave **inserta una entrada** si la clave no existe y sustituye su valor si ya existe [@psf314asignacion]:

````python
existencias["P03"] = 3      # inserción de un nuevo elemento
existencias["P01"] = 15     # sustituye un valor
print(existencias)
````

- Un **diccionario es mutable**.
- Su **recorrido conserva el orden de inserción** (desde Python 3.7), que no implica clasificación alfabética o numérica.
- Actualizar una clave no cambia su posición, eliminarla e insertarla de nuevo la sitúa al final [@psf314diccionarios].

## Consulta y operaciones básicas

La consulta de un diccionario permite recuperar información sin modificar sus entradas. Las operaciones básicas permiten obtener el valor asociado a una clave, comprobar si una clave está presente o ausente, determinar cuántas entradas contiene el diccionario y eliminar entradas.

- **Obtener un valor** asociado a una clave:

````python
existencias["P01"]
````

Al intentar acceder al valor asociado con una clave que no existe en el diccionario, se genera una excepción de tipo `KeyError`.

- **Comprobar la existencia** de una clave en un diccionario:

````python
"P02" in existencias
````

````python
"P04" not in existencias
````

- **Obtener la cantidad** de elementos:

````python
len(existencias)
````

- **Eliminar una entrada**:

````python
del existencias['P02']
````

## Iteración sobre diccionarios

El recorrido directo de un diccionario produce claves. Los métodos `keys()`, `values()` e `items()` ofrecen vistas dinámicas de claves, valores y pares, respectivamente. [@psf314diccionarios].

- El método `keys()` permite iterar **sobre las claves** del diccionario:

````python
existencias = {"P01": 12, "P02": 0}
for codigo in existencias.keys():
    print(codigo)
````

- El método `values()` permite iterar **sobre valores** del diccionario:

````python
existencias = {"P01": 12, "P02": 0}
for codigo in existencias.values():
    print(codigo)
````

- El método `items()`, por medio del **desempaquetado** asigna los componentes de una secuencia a varios identificadores. Esta operación permite utilizar los pares (`clave: valor`) directamente en un recorrido [@psf314asignacion].


````python
existencias = {"P01": 12, "P02": 0}

for codigo, cantidad in existencias.items():
    print(codigo, cantidad)
````

## Cambios durante el recorrido

Antes de ejecutar el siguiente código, analice qué operaciones realiza y qué resultado se espera obtener:

````python
existencias = {"P01": 12, "P02": 0, "P03": 5}

for codigo, cantidad in existencias.items():
    if cantidad == 0:
        del existencias[codigo]
````

Agregar o eliminar entradas durante un recorrido sobre una vista puede producir `RuntimeError` o impedir que se visiten todas las entradas [@psf314diccionarios]. Para evitar esa dependencia, **un algoritmo que elimine valores debe separar la selección de claves de su eliminación.**



:::{note} Ejercicio

Diseñe e implemente un procedimiento que elimine de un diccionario los elementos cuyos valores sean cero.

Pruebe la función con un diccionario vacío, otro sin valores con cero y otro con varios valores que deban eliminarse.

:::

**Solucion:**

````python
def elimina_valores(diccionario, valor=0):

    # selección de valores objetivos
    claves_a_eliminar = []
    for clave, valor in diccionario.items():
        if valor == 0:
            claves_a_eliminar.append(clave)

    # eliminación de valores detectados
    for clave in claves_a_eliminar:
        del existencias[clave]

existencias = {"P01": 12, "P02": 0, "P03": 0}
elimina_valores(existencias)

print(existencias)
````

1. Especificación de la función:

| Elemento                | Especificación                                                                                                                                                                                                                         |
|:------------------------|:---------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------|
| **Propósito**           | Contar las apariciones de cada palabra en una lista.                                                                                                                                                                                   |
| **Parámetros**          | `palabras`: lista de cadenas que se desea procesar.                                                                                                                                                                                    |
| **Precondición**        | `palabras` es una lista finita y todos sus elementos son cadenas no vacías. La lista puede estar vacía.                                                                                                                                |

2. Implementación:

````python
def contar_frecuencias(palabras):

    frecuencias = {}
    for palabra in palabras.keys():
        if palabra in frecuencias:
            frecuencias[palabra] = frecuencias[palabra] + 1
        else:
            frecuencias[palabra] = 1

    return frecuencias
````

## Asignación y copia

La asignación de un diccionario a otro identificador no crea una copia, ambos nombres se refieren al mismo objeto (**asignación de referencias**).

El método `copy()` crea otro diccionario mediante una **copia superficial**, es decir, se reproducen las entradas, pero conserva referencias a los objetos contenidos [@psf314copia].

En el siguiente ejemplo, se puede distinguir el efecto:

````python
original = {"P01": 12}
mismo_original = original

copia = original.copy()
mismo_original["P01"] = 4

print(original)
print(copia)
````

## Valores mutables compartidos

Una copia superficial no duplica las listas internas. Por esta razón, modificar una lista compartida resulta observable desde ambos diccionarios [@psf314copia].

Para examinar esta relación, observe el efecto en la ejecucución del siguiente código:

````python
notas = {"Cálculo": [7, 5], "Física": []}
copia = notas.copy()
copia["Física"].append(4)

print(notas["Física"])
print(copia["Física"])
````

En consecuencia, **modificar un objeto compartido y sustituir una asociación son operaciones diferentes**.

Para modificar independientemente las listas de notas, el algoritmo debe crear otro diccionario y copiar cada lista. Para el ejemplo anterior, basta con dos capas:

````python
notas = {"Cálculo": [7, 5], "Física": []}

# crea copia independiente de los valores mutables (listas) del diccionario
copia_independiente = {}
for asignatura, calificaciones in notas.items():
    copia_independiente[asignatura] = calificaciones.copy()

copia_independiente["Física"].append(4)
print(notas["Física"])
print(copia_independiente["Física"])
````

## Diccionarios anidados y listas de diccionarios

Las colecciones pueden contener otras colecciones [@psf314estructuras]. La composición debe responder a las consultas previstas.

Por ejemplo, si se desea consultar por claves para nombre como notas, cada valor puede ser un **diccionario de atributos**:

````python
asignaturas = {
    "457000": {
        "nombre": "Inglés instrumental I",
        "creditos": 3,
        "notas": [7, 5]
    },
    "457001": {
        "nombre": "Inglés instrumental II",
        "creditos": 3,
        "notas": []
    },
}
print(asignaturas["457000"]["notas"])
````

La primera clave selecciona la asignatura y la segunda selecciona uno de sus atributos.

Por otra parte, si el propósito es conservar una secuencia de registros, puede utilizarse una **lista de diccionarios**:

````python
asignaturas = [
    {
        "codigo": "457000",
        "nombre": "Inglés instrumental I",
        "creditos": 3,
        "notas": [7, 8]
    },
    {
        "codigo": "457001",
        "nombre": "Inglés instrumental II",
        "creditos": 3,
        "notas": []
    },
]
print(asignaturas[0]["codigo"])
````

La posición selecciona un registro y la clave selecciona un atributo. Si se requiere localizar un registro por código, esta representación exige diseñar una búsqueda.
