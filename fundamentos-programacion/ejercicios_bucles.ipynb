BCL-001. Diseñar un programa que muestra la __tabla de multiplicar__ (del 1 hasta el 10) de cualquier número ingresado por el usuario, por ejemplo,

```
Ingrese un numero: 4
Tabla del 4
4 x 1 = 4
4 x 2 = 8
4 x 3 = 12
:
.
4 x 10 = 40
```

Otro ejemplo,

```
Ingrese un numero: 7
Tabla del 7
7 x 1 = 7
7 x 2 = 14
7 x 3 = 21
:
.
7 x 10 = 70
```

Solución (código fuente).

````python
n = int(input('Ingrese un número: '))

print('Tabla del ', n)
i = 1
while i <= 10:
    resultado = n * i
    print(n, 'x', i, '=', resultado)
    i += 1
````

BCL-002. Diseñar un programa que muestre las 12 primeras tablas de multiplicar (1 al 12).

Solución.

````python
for j in range(1,13,1):
    print('Tabla del', j)
    for i in range(1,13,1):
        resultado = j * i
        print(j, 'x', i, '=', resultado)
````

BCL-005. Diseñe un programa que muestre los __números primos__ que existen entre dos números ingresados.

```
Número menor: 1
Número mayor: 20
2
3
5
7
11
13
17
19
```

Solución (código fuente).

````python
menor = int(input('menor? '))
mayor = int(input('mayor? '))

while menor < (mayor-1):
    menor += 1
    divisor = menor - 1
    cuenta = 0

    while divisor > 1:
        resto = menor % divisor
        if resto == 0:
            cuenta += 1

        divisor -= 1

    if cuenta == 0:
        print(menor)
````

BCL-008. El procedimiento para encontrar la __representación binaria__ de un número entero positivo (decimal), se describe a continuación:
* Divida el número decimal por 2 y registre el resto.
* Si el cociente de la división no es cero (0), vuelva al paso anterior. De lo contrario, la reresentación binaria del número decimal estará conformada por todos los restos de las divisiones sucesivas, escritos orden inverso al que fueron registrados.

Por ejemplo, a continuación se muestra el proceso aplicado para obtener la representación binaria del número 130:

\begin{align}
    130 &\div 2 = 65 &\longrightarrow r = 0\\
    65 &\div 2 = 32 &\longrightarrow r = 1\\
    32 &\div 2 = 16 &\longrightarrow r = 0\\
    16 &\div 2 = 8 &\longrightarrow r = 0\\
    8 &\div 2 = 4 &\longrightarrow r = 0\\
    4 &\div 2 = 2 &\longrightarrow r = 0\\
    2 &\div 2 = 1 &\longrightarrow r = 0\\
    1 &\div 2 = 0 &\longrightarrow r = 1
\end{align}

En consecuencia, la representación binaria de 130 es `10000010`.
