# Ejercicio: Reclutamiento

El archivo de texto <a href="./datos/puntajes_2018.txt" download="puntajes_2018.txt">`puntajes_2018.txt`</a> contiene los puntajes de un conjunto de personas que postularon a un cargo administrativo en una empresa de reclutamiento de profesionales. Cada registro contiene un puntaje entero.

Puede diseñar funciones auxiliares con responsabilidades claras, cada una destinada a una tarea específica y ser reutilizable. Para cada ejercicio, identifique, diseñe y ejecute pruebas.

1. Considerando que la empresa dispone de 25 vacantes, diseñe la función `seleccionar_postulantes(nombre_archivo)` que crea los siguientes archivos:

   - `mejores_puntajes.txt`: con los 25 mejores puntajes, correspondientes a las personas seleccionadas para ocupar las vacantes.
   - `lista_de_espera.txt`: con los 5 mejores puntajes restantes, correspondientes a las personas que quedarán en la lista de espera.

2. Diseñe la función `promedio(nombre_archivo)`, que devuelva el promedio de los puntajes contenidos en el archivo de texto.

3. Diseñe la función `menores_que(nombre_archivo, umbral)`, que devuelva una lista con todos los puntajes del archivo de texto que sean menores que el umbral recibido como argumento.
