# Ejercicios: tratamiento de archivos

**TA-001.** El archivo de texto <a href="puntajes_2018.txt" download="puntajes_2018.txt">`puntajes_2018.txt`</a> contiene los puntajes de un conjunto de personas que postularon a un cargo administrativo en una empresa de reclutamiento de profesionales. Cada registro contiene un puntaje entero.

Puede diseñar funciones auxiliares con responsabilidades claras, cada una destinada a una tarea específica y ser reutilizable. Para cada ejercicio, identifique, diseñe y ejecute pruebas.

1. Considerando que la empresa dispone de 25 vacantes, diseñe la función `seleccionar_postulantes(nombre_archivo)` que crea los siguientes archivos:

   - `mejores_puntajes.txt`: con los 25 mejores puntajes, correspondientes a las personas seleccionadas para ocupar las vacantes.
   - `lista_de_espera.txt`: con los 5 mejores puntajes restantes, correspondientes a las personas que quedarán en la lista de espera.
2. Diseñe la función `promedio(nombre_archivo)`, que devuelva el promedio de los puntajes contenidos en el archivo de texto.
3. Diseñe la función `menores_que(nombre_archivo, umbral)`, que devuelva una lista con todos los puntajes del archivo de texto que sean menores que el umbral recibido como argumento.


**TA-002.** Un condominio residencial sospecha que existe una fuga en su sistema de riego automático de áreas verdes. Para investigarlo, se instala un medidor que registra el consumo de agua en intervalos consecutivos de 10 minutos, mientras el riego permanece apagado. El consumo de cada intervalo, expresado en litros, se almacena en una línea del archivo <a href="consumos.txt" download="consumos.txt">`consumos.txt`</a>.

Desarrolle un programa que:

1. Solicite un umbral de consumo sospechoso.
2. Lea todas las mediciones.
3. Determine la cantidad de intervalos, el consumo total y el consumo promedio.
4. Cuente los intervalos cuyo consumo supera estrictamente el umbral.
5. Identifique el mayor consumo y el número del intervalo en que ocurrió. Si se repite, informe el primero.
6. Muestre los resultados en pantalla.

Para la solución, se sugiere la siguiente estructura modular:
- `leer_consumos(nombre_archivo):` abre el archivo indicado, lee las mediciones y las almacena en una lista de números reales. Cierra el archivo y devuelve la lista. Debe controlar el caso en que el archivo no exista.

- `analizar_consumos(consumos, umbral):` recibe la lista de mediciones y el umbral. Calcula la cantidad de intervalos, el consumo total, el promedio y la cantidad de intervalos que superan estrictamente el umbral. Además, identifica el consumo máximo y el primer intervalo en que ocurrió, numerando desde 1. Devuelve los resultados obtenidos.

- `mostrar_resultados(...):` recibe los resultados del análisis y los presenta en pantalla con mensajes que permitan interpretar cada valor.

- `procesar_consumos():` coordina la ejecución del programa: solicita el umbral, llama a la función de lectura y, si existen mediciones, realiza el análisis y muestra los resultados mediante las funciones correspondientes. Si el archivo está vacío, informa que no existen mediciones.

**TA-003.**
Un refrigerador de alimentos registra su temperatura cada cinco minutos. Una lectura elevada aislada requiere revisión, mientras que varias lecturas elevadas consecutivas pueden indicar una falla. Las mediciones se almacenan en el archivo <a href="temperaturas.txt" download="temperaturas.txt">`temperaturas.txt`</a>, con una temperatura por línea, en el orden en que fueron registradas.

Considere que una temperatura superior a 8°C se encuentra fuera del rango establecido.

Desarrolle un programa que:

1. Cuente las lecturas fuera de rango.
2. Identifique los episodios de temperaturas consecutivas fuera de rango.
3. Determine el inicio, el final y la cantidad de lecturas de cada episodio.
4. Identifique el episodio más largo. En caso de empate, seleccione el primero de ellos.
5. Genere el archivo `alertas.txt` con una descripción de cada episodio. Un ejemplo de la estructura del archivo se muestra a continuación:
    ```text
    INFORME DE ALERTAS DE TEMPERATURA
    Umbral: 8 °C
    Total de lecturas: 8
    Lecturas fuera de rango: 5
    Cantidad de episodios: 2

    EPISODIO 1
    Lectura inicial: 2
    Lectura final: 3
    Cantidad de lecturas: 2

    EPISODIO 2
    Lectura inicial: 5
    Lectura final: 7
    Cantidad de lecturas: 3

    EPISODIO MÁS LARGO
    Número de episodio: 2
    Lectura inicial: 5
    Lectura final: 7
    Cantidad de lecturas: 3
    ```

Los números de lectura indican su posición en el archivo de entrada, comenzando desde 1.

**TA-004.**
Una organización no gubernamental (ONG) lanzó una campaña en redes sociales y desea conocer las reacciones de los usuarios. Para esto, cuenta con un archivo unificado con todos los mensajes publicados en cada red social y necesita clasificarlos automáticamente según la presencia de palabras positivas y negativas.

El archivo <a href="mensajes.txt" download="mensajes.txt">`mensajes.txt`</a> contiene un mensaje por línea. Para simplificar el procesamiento, todos están escritos en minúsculas, sin tildes, signos de puntuación, emojis ni enlaces. Las palabras pueden estar separadas por uno o más espacios.

Considere las siguientes palabras para la clasificación:

| Sentimiento | Palabras                                                    |
|:------------|:------------------------------------------------------------|
| Positivo    | `bueno`, `buena`, `excelente`, `genial`, `encanta`, `feliz` |
| Negativo    | `malo`, `mala`, `terrible`, `pesimo`, `odio`, `triste`      |

Cada aparición de una palabra positiva suma un punto y cada aparición de una palabra negativa resta un punto. Las demás palabras no modifican el puntaje.

El mensaje se clasifica como:

- **Positivo:** puntaje mayor que cero.
- **Negativo:** puntaje menor que cero.
- **Neutro:** puntaje igual a cero, tanto si no contiene palabras clave como si las apariciones positivas y negativas se compensan.

La comparación debe identificar palabras completas. Por ejemplo, `buena` no debe coincidir con `buenas`. Si una palabra aparece varias veces, se contabilizan todas sus apariciones.

Desarrolle un programa que:

1. Lea y analice los mensajes.
2. Calcule el puntaje de cada mensaje y determine su clasificación.
3. Genere tres archivos: `positivos.txt`, `negativos.txt` y `neutros.txt`.
4. Escriba cada mensaje en el archivo correspondiente, conservando su texto y orden de aparición. Cada mensaje debe escribirse una sola vez.
5. Muestre en pantalla: la cantidad total de mensajes analizados; la cantidad y el porcentaje de mensajes de cada categoría; el número y el puntaje del mensaje más positivo; el número y el puntaje del mensaje más negativo. Por ejemplo:
    ```text
    Mensajes analizados: 5
    Positivos: 2 (40.0 %)
    Negativos: 1 (20.0 %)
    Neutros: 2 (40.0 %)
    Mensaje mas positivo: 1, puntaje: 2
    Mensaje mas negativo: 2, puntaje: -2
    ```

Los mensajes se numeran desde 1. Para identificar el más positivo, considere solamente aquellos con puntaje mayor que cero. Para el más negativo, aquellos con puntaje menor que cero. En caso de empate, informe el primero. Si una categoría no tiene mensajes, indíquelo.

Los archivos de salida deben generarse nuevamente en cada ejecución del programa, incluso si alguna categoría queda vacía. Si el archivo de entrada está vacío, debe informarse que no hay mensajes para analizar, sin calcular porcentajes. Si no existe, debe mostrarse un mensaje y finalizar el procesamiento.

**TA-005.**
Dos sensores midieron precipitaciones, expresadas em milímetros de agua caída, en un mismo lugar. Ambos presentan interrupciones, pero no necesariamente durante las mismas horas. Se necesita construir una serie que aproveche las observaciones disponibles e identificar las discrepancias entre ambos sensores.

Los archivos <a href="sensor_a.txt" download="sensor_a.txt">`sensor_a.txt`</a> y <a href="sensor_b.txt" download="sensor_b.txt">`sensor_b.txt`</a> contienen un valor por línea para cada hora de una jornada. Ambos tienen exactamente 24 líneas y la línea correspondiente a una misma posición representa la misma hora.

- Un valor mayor o igual que cero representa una medición válida en milímetros.
- El valor `-999` indica que no existe medición.

Diseñe un programa que solicite al usuario un valor de tolerancia (en milímetros) e integre las mediciones de ambos sensores aplicando los siguientes criterios:

| Situación                                                           | Valor de la serie combinada       |
|:--------------------------------------------------------------------|:----------------------------------|
| Ambos sensores tienen datos y su diferencia no supera la tolerancia | Promedio de ambos valores         |
| Solo un sensor tiene datos                                          | Valor del sensor disponible       |
| Ninguno tiene datos                                                 | `-999`                            |
| Ambos tienen datos, pero su diferencia supera la tolerancia         | `-999` y registro de discrepancia |

Además, el programa debe generar:
1. El archivo `precipitacion_integrada.txt`, con 24 valores, uno por línea.
2. El archivo `discrepancias.txt` debe incluir el encabezado en la primera línea y una fila por discrepancia, con los siguientes campos separados por punto y coma: la hora (`00` a `23`), las mediciones de ambos sensores y su diferencia absoluta. Solo se incluyen las horas en que ambas mediciones son válidas y la diferencia supera la tolerancia. Por ejemplo:
    ```text
    hora;sensor_a_mm;sensor_b_mm;diferencia_mm
    04;1.0;5.0;4.0
    11;2.5;4.0;1.5
    18;3.0;6.0;3.0
    ```

3. Un resumen en pantalla con la cantidad de horas resueltas y pendientes. Como por ejemplo,
    ```text
    RESUMEN DEL PROCESAMIENTO
    Tolerancia: 1.0 mm
    Horas procesadas: 24

    Horas resueltas: 18
      Con ambas mediciones: 12
      Con una medición: 6

    Horas pendientes: 6
      Sin datos: 3
      Discrepancias: 3

    Archivos generados:
    precipitacion_combinada.txt
    discrepancias.txt
    ```

Descripción de los items del reporte:

| Item                             | Descripción                                                                                              |
|:---------------------------------|:---------------------------------------------------------------------------------------------------------|
| Tolerancia                       | Diferencia máxima permitida entre las mediciones de ambos sensores para calcular el promedio.            |
| Horas procesadas                 | Cantidad de horas analizadas.                                                                            |
| Horas resueltas                  | Horas para las que se obtuvo un valor válido de precipitación.                                           |
| Con ambas mediciones             | Ambos sensores tienen datos válidos y su diferencia no supera la tolerancia. Se utiliza el promedio.     |
| Con una medición                 | Un sensor tiene un dato válido y el otro registra `-999`. Se utiliza el dato válido.                     |
| Horas pendientes                 | Horas para las que no fue posible obtener un valor confiable. Se escribe `-999` en el archivo combinado. |
| Sin datos                        | Ambos sensores registran `-999`.                                                                         |
| Discrepancias                    | Ambos sensores tienen datos válidos, pero su diferencia supera la tolerancia.                            |
