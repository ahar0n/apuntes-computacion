En Python, La comparación de cadenas se realiza evaluando el contenido carácter por carácter mediante los puntos de código numéricos del estándar Unicode, de izquierda a derecha... [mas información](https://docs.python.org/es/3.15/howto/unicode.html)
- Los comparadores, __comparan el valor numérico del código Unicode__ del primer carácter de cada cadena. Si son iguales, avanza al siguiente carácter hasta hallar una diferencia.
- Las letras __mayúsculas__ tienen un valor numérico Unicode __menor que las minúsculas__, por lo que en las reglas de ordenación estricta de Python las mayúsculas se consideran "menores" y aparecen antes que las minúsculas.
